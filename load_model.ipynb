```python
import joblib
import inspect

pipeline = joblib.load(MODEL_PATH)


print(f"Тип модели: {type(pipeline).__name__}")
print(f"Количество признаков: {pipeline.n_features_in_}")


if hasattr(pipeline, 'feature_names_in_'):
    print(f"\nНазвания признаков на входе ({len(pipeline.feature_names_in_)} шт):")
    print(pipeline.feature_names_in_)


print(f"Шаги pipeline:")
for step_name, step in pipeline.named_steps.items():
    print(f"  {step_name}: {type(step).__name__}")


print(f"Детали препроцессора:")
preprocessor = pipeline.named_steps['preprocessor']

for name, transformer, columns in preprocessor.transformers_:
    print(f"\n {name}:")
    print(f"    Трансформер: {type(transformer).__name__}")
    print(f"    Колонки ({len(columns)} шт): {list(columns)}")
    if hasattr(transformer, 'categories_'):
        print(f"    Категории: {transformer.categories_}")


print(f"Параметры модели (CatBoost):")
model = pipeline.named_steps['model']
print(f"  {type(model).__name__}")
params = model.get_params()
for key, value in params.items():
    if value is not None:
        print(f"    {key}: {value}")


print(f"\nПризнаки после препроцессора:")
if hasattr(preprocessor, 'get_feature_names_out'):
    output_features = preprocessor.get_feature_names_out()
    print(f"  Количество: {len(output_features)}")
    print(f"  Названия: {output_features}")
```
